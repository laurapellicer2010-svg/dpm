###### **Relevant notes**
- Do not edit locked cells, unless you need to add a new import. Session Configuration tabs and Engine Execution tabs should remain as is.
- The notebook can be run again in any environment: it leaves the tables equal to the values written here, without duplicating rows.
- **Reference tables** (MERGE): a new row is inserted and a row whose value changed is updated. Rows that are not listed here are not deleted; to remove one, run a DELETE.
    - ref_tmplt_csl_perim: FRP_TPMAPA_MAPEAMENTO of the previous solution (65 rows) plus the subtemplates 04.02.1, 04.02.2, 04.03.1 and 04.04.1, with the perimeter of their parent template.
    - ref_tmplt_ctry: FRP_ZAXIX_MAPA_COM_PAIS of the previous solution.
    - ref_src_datapoint_typ: datapoint type of each source table (Source Table of the Sources tab).
    - The template is written as in the Catalog tab, without the letter of the framework (FINREP, 20.04).
- **Issue catalogue** (MERGE): the list contains every code of the deployed engine version (DPM_Issue_Catalog.xlsx). A code that is not in the table yet is inserted as a current version (ctrl_state = 1); codes already in the table are not changed. To add a code, add it to the list (and to DPM_Issue_Catalog.xlsx) and run the notebook again. Changing or removing an existing code is not covered by this notebook.
- To change a value of a reference table, edit it in this notebook and run it again.

##### **[Consolidation - Datapoint type by source table]**

```python
%%sql
MERGE INTO dp_lgl_reg_dpm_ref_src_datapoint_typ AS t
USING (
    SELECT
        src_tbl_id
        , datapoint_typ
    FROM VALUES
    ('DP_LGL_REG_ACG_CSL_LDGR_1', 'CONTABILISTICO')
    , ('DP_LGL_REG_FACT_SCRTS_LAST', 'OPERACIONAL')
    , ('FP_M_BALANCETE_IFRS', 'CONTABILISTICO')
    , ('FP_M_FINREP', 'OPERACIONAL')
    AS v (src_tbl_id, datapoint_typ)
) AS s
    ON UPPER(TRIM(t.src_tbl_id)) = UPPER(TRIM(s.src_tbl_id))
WHEN MATCHED AND NOT (t.datapoint_typ <=> s.datapoint_typ) THEN UPDATE SET datapoint_typ = s.datapoint_typ
WHEN NOT MATCHED THEN INSERT (src_tbl_id, datapoint_typ)
    VALUES (s.src_tbl_id, s.datapoint_typ)
```
