###### **Relevant notes**
- Do not edit locked cells, unless you need to add a new import. Session Configuration tabs and Engine Execution tabs should remain as is.
- The notebook can be run again in any environment: it leaves the tables equal to the values written here, without duplicating rows.
- **Reference tables** (MERGE): a new row is inserted and a row whose value changed is updated. Rows that are not listed here are not deleted; to remove one, run a DELETE.
    - ref_tmplt_csl_perim: FRP_TPMAPA_MAPEAMENTO of the previous solution (65 rows) plus the subtemplates 04.02.1, 04.02.2, 04.03.1 and 04.04.1, with the perimeter of their parent template.
    - ref_tmplt_ctry: FRP_ZAXIX_MAPA_COM_PAIS of the previous solution.
    - ref_src_datapoint_typ: datapoint type of each source table (Source Table of the Sources tab).
    - The template is written as in the Catalog tab, without the letter of the framework (FINREP, 20.04).
- **Issue catalogue** (versioned, SCD 2.2): the list must contain every code of the deployed engine version (DPM_Issue_Catalog.xlsx). A changed code closes its current version and opens a new one; a code that is no longer in the list gets a logically deleted version (row_present_flg = 0); unchanged codes keep their version.
- To change a value, edit it in this notebook (and in DPM_Issue_Catalog.xlsx for the issue catalogue) and run it again.

##### **[Consolidation - Templates broken down by country]**

##### **[Consolidation - Datapoint type by source table]**

```python
%%sql
MERGE INTO dp_lgl_reg_dpm_ref_src_datapoint_typ AS t
USING (
    SELECT
        src_tbl_id
        , datapoint_typ
    FROM VALUES
    ('DP_LGL_REG_ACG_CSL_LDGR_1', 'CONTABILISTICO')
    , ('DP_LGL_REG_FACT_SCRTS_LAST', 'OPERACIONAL')
    , ('FP_M_BALANCETE_IFRS', 'CONTABILISTICO')
    , ('FP_M_FINREP', 'OPERACIONAL')
    AS v (src_tbl_id, datapoint_typ)
) AS s
    ON UPPER(TRIM(t.src_tbl_id)) = UPPER(TRIM(s.src_tbl_id))
WHEN MATCHED AND NOT (t.datapoint_typ <=> s.datapoint_typ) THEN UPDATE SET datapoint_typ = s.datapoint_typ
WHEN NOT MATCHED THEN INSERT (src_tbl_id, datapoint_typ)
    VALUES (s.src_tbl_id, s.datapoint_typ)
```

```python
# The list of the previous cell becomes the current issue catalogue (SCD 2.2, key iss_cd),
# with the same versioning as the parameterisation tables.
issue_catalog_table = "dp_lgl_reg_dpm_param_iss_ctlg"
issue_catalog_source = "dpm_iss_ctlg_src"

ISSUE_KEY = "iss_cd"
ISSUE_COLUMNS = [
    "iss_cd", "dmn_cd", "stg_cd", "pattern_cd", "sev_cd", "scp_cd",
    "msg_tmplt_ln", "param_nm_list_ln", "act_flg",
]
PRESENT_COLUMN = "row_present_flg"
CURRENT = 1
PRESENT = 1
DELETED = 0
CLOSE = "CLOSE"
SEEN = "SEEN"


def content_hash(frame):
    """Hash of the content of a catalogue row, including whether the code is present."""
    return F.sha2(F.to_json(F.struct(*[F.col(name) for name in ISSUE_COLUMNS + [PRESENT_COLUMN]])), 256)


def plan_issue_catalog(target, incoming, version_ts):
    """Changes that leave the current issue catalogue equal to the list.

    Returns the actions on the current versions, by surrogate key (CLOSE when the code
    changed or is no longer in the list, SEEN when it is unchanged), and the new versions:
    new or changed codes, and a logically deleted version (row_present_flg = 0) for the
    codes that are no longer in the list.
    """
    key_column = next(name for name in target.columns if name.startswith("sk_"))
    current = target.where(F.col("ctrl_state") == CURRENT)
    removed = (
        current.where(F.col(PRESENT_COLUMN) == PRESENT)
        .join(incoming.select(ISSUE_KEY), ISSUE_KEY, "left_anti")
        .select(*ISSUE_COLUMNS)
        .withColumn(PRESENT_COLUMN, F.lit(DELETED))
    )
    wanted = incoming.select(*ISSUE_COLUMNS).withColumn(PRESENT_COLUMN, F.lit(PRESENT)).unionByName(removed)
    wanted = wanted.withColumn("_hash", content_hash(wanted))
    current_hash = current.select(key_column, ISSUE_KEY, content_hash(current).alias("_current_hash"))
    joined = wanted.join(current_hash, ISSUE_KEY, "left")
    unchanged = F.col("_hash") == F.col("_current_hash")

    actions = joined.where(F.col(key_column).isNotNull()).select(
        F.col(key_column).alias("row_sk"),
        F.when(unchanged, F.lit(SEEN)).otherwise(F.lit(CLOSE)).alias("act_cd"),
    )
    version = F.lit(version_ts).cast("timestamp")
    technical = {
        "dpm_vrs_dt": version,
        "dpm_ingestion_dt": version,
        "ref_start_dt": version,
        "ref_last_dt": version,
        "ctrl_start_dt": version,
        "ctrl_updt_dt": version,
        "ctrl_state": F.lit(CURRENT),
    }
    last_key = target.agg(F.max(key_column)).first()[0] or 0
    changed = joined.where(F.col("_current_hash").isNull() | ~unchanged).withColumn(
        key_column, F.lit(last_key) + F.row_number().over(Window.orderBy(ISSUE_KEY))
    )
    kept = set(ISSUE_COLUMNS + [PRESENT_COLUMN, key_column])
    new_versions = changed.select(*[
        (F.col(field.name) if field.name in kept else technical.get(field.name, F.lit(None)))
        .cast(field.dataType).alias(field.name)
        for field in target.schema.fields
    ])
    # Fixed before the table is written: the changes are computed on its current content.
    return actions.localCheckpoint(eager=True), new_versions.localCheckpoint(eager=True)


def apply_issue_catalog(actions, new_versions, version_ts):
    """Closes the versions that changed, marks the unchanged ones as seen and inserts the new ones."""
    key_column = next(name for name in new_versions.columns if name.startswith("sk_"))
    version = F.lit(version_ts).cast("timestamp")
    (
        DeltaTable.forName(spark, issue_catalog_table)
        .alias("t")
        .merge(actions.alias("s"), F.col(f"t.{key_column}") == F.col("s.row_sk"))
        .whenMatchedUpdate(
            condition=F.col("s.act_cd") == CLOSE,
            set={"ref_end_dt": version, "ctrl_end_dt": version, "ctrl_updt_dt": version, "ctrl_state": F.lit(0)},
        )
        .whenMatchedUpdate(set={"ref_last_dt": version, "ctrl_updt_dt": version})
        .execute()
    )
    new_versions.write.format("delta").mode("append").saveAsTable(issue_catalog_table)


incoming = spark.table(issue_catalog_source)
duplicated = [
    row[ISSUE_KEY]
    for row in incoming.groupBy(ISSUE_KEY).agg(F.count(F.lit(1)).alias("row_nr")).where(F.col("row_nr") > 1).toLocalIterator()
]
if duplicated:
    raise ValueError(f"Issue codes repeated in the list: {sorted(duplicated)}")

version_ts = datetime.now(timezone.utc).replace(tzinfo=None)
actions, new_versions = plan_issue_catalog(spark.table(issue_catalog_table), incoming, version_ts)
apply_issue_catalog(actions, new_versions, version_ts)
action_nr = {row["act_cd"]: row["row_nr"] for row in actions.groupBy("act_cd").agg(F.count(F.lit(1)).alias("row_nr")).toLocalIterator()}
print(
    f"{issue_catalog_table}: {new_versions.count()} versions inserted, "
    f"{action_nr.get(CLOSE, 0)} closed, {action_nr.get(SEEN, 0)} unchanged."
)
```
